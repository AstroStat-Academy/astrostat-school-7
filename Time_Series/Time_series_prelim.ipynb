## Missing values

Real time series — *especially* in astronomy — are rarely complete. Gaps arise from:

- **Observing constraints:** daytime, weather, the target being below the horizon, the Moon being too close, the satellite being in the South Atlantic Anomaly.
- **Instrumental issues:** detector saturation, cosmic-ray hits, telemetry dropouts, scheduled downtime.
- **Irregular cadence by design:** queue-scheduled observations, target-of-opportunity interrupts.

The result is a series $\{x_t\}$ where some $t \in T_0$ have no observation. This is a problem because most classical time series tools (ACF estimation, ARMA fitting, FFT-based spectral analysis) assume evenly-spaced, complete data.

**Approaches to handling missing data:**

- **Ignore and work irregularly:** use methods that natively handle uneven sampling (Lomb–Scargle periodogram, structure functions, Gaussian processes).
- **Interpolation:** fill in gaps with a deterministic rule (linear, spline, nearest neighbor). Simple, but introduces correlations that aren't really there and underestimates uncertainty in the gaps.
- **Model-based imputation:** treat the missing values as latent variables and infer them under a probabilistic model. This is the principled approach — it propagates uncertainty correctly and lets you do inference on the parameters and the gaps jointly.

**Gaussian processes (preview):** one particularly powerful model-based approach is the **Gaussian process** (GP), which treats the entire time series as a draw from a joint Gaussian distribution whose covariance structure is set by a kernel function $k(t, t')$. Given the observed points, the conditional distribution of the unobserved points is *also* Gaussian, with closed-form mean and variance — so a GP naturally "fills in" gaps with a best-estimate curve *and* a principled uncertainty band that widens inside long gaps and shrinks near data. We'll come back to GPs later in the school; for now, just keep in mind that the missing-data problem is one of their natural application areas.

## Objective of time series analysis

The goal is to set up a **model** $\{X_t\}$ — a specification of the joint distribution of the underlying random variables — that:

1. **Describes** the observed series $\{x_t\}$, capturing its key features (trend, seasonality, autocorrelation structure, variance behavior).
2. **Predicts** future values $X_{t+1}, X_{t+2}, \ldots$ with quantified uncertainty.
3. **Tests hypotheses** about the underlying process — e.g., is there a real trend in global mean temperature, or is the apparent rise consistent with stationary noise? Is the variability in this AGN light curve periodic, or red-noise consistent with a damped random walk?

A good model is the bridge between a finite, noisy realization $\{x_t\}$ and inference about the stochastic process that produced it.

## Frequency-domain analysis: the FFT

So far we've described a stationary time series through its **autocovariance function** $\gamma(h)$ — a function of time-lag $h$, living in the *time domain*. There is an equivalent, complementary description in the *frequency domain*: instead of asking "how correlated are observations separated by lag $h$?", we ask "how much of the variance is carried by oscillations at frequency $\omega$?".

The bridge between the two descriptions is the **Fourier transform**, and in practice we compute it with the **Fast Fourier Transform (FFT)** — an $\mathcal{O}(N \log N)$ algorithm for what would naively be an $\mathcal{O}(N^2)$ computation. This is the workhorse of practical time series analysis and is everywhere in astronomy.

### The Fourier transform of a finite series

Given an evenly-spaced time series $x_0, x_1, \ldots, x_{N-1}$ sampled at interval $\Delta t$, the **discrete Fourier transform (DFT)** is

$$\tilde{x}_k = \sum_{t=0}^{N-1} x_t\, e^{-2\pi i\, k t / N}, \quad k = 0, 1, \ldots, N-1.$$

Each $\tilde{x}_k$ is a complex number representing the amplitude and phase of the oscillation at frequency

$$\omega_k = \frac{k}{N \Delta t}, \quad k = 0, 1, \ldots, N/2.$$

The transform is invertible: given the $\tilde{x}_k$ you can reconstruct the $x_t$ exactly. So no information is lost — the time-domain and frequency-domain representations are two views of the same data.

### Key frequencies

Two frequencies define the resolution of the analysis:

- **Nyquist frequency:** $\omega_{\rm Ny} = 1/(2\Delta t)$. The highest frequency you can resolve with sampling interval $\Delta t$. Power above $\omega_{\rm Ny}$ gets *aliased* — folded back into lower frequencies — and is irretrievably confused with the resolved band. This is the fundamental sampling-theorem constraint.
- **Frequency resolution:** $\Delta\omega = 1/(N \Delta t) = 1/T$, where $T = N \Delta t$ is the total observation duration. Longer baselines give finer frequency resolution. Two oscillations closer than $\Delta\omega$ apart cannot be separated.

The practical implication: to detect a signal of frequency $\omega$, you need $\omega < \omega_{\rm Ny}$ (sample fast enough) **and** you need $T \gtrsim$ several periods (observe long enough).

### The periodogram

The standard frequency-domain summary of a time series is the **periodogram**:

$$I(\omega_k) = \frac{1}{N} \left| \tilde{x}_k \right|^2.$$

It is a non-negative, real-valued estimate of how much variance sits at each frequency. The total variance decomposes as

$$\frac{1}{N} \sum_{t=0}^{N-1} (x_t - \bar{x})^2 \;\approx\; \sum_{k=1}^{N/2} I(\omega_k) \cdot \Delta\omega,$$

so the area under the periodogram equals the variance of the series — this is **Parseval's theorem**. The periodogram is the empirical analog of the **power spectral density (PSD)** $f(\omega)$, which is the Fourier transform of the autocovariance:

$$f(\omega) = \sum_{h=-\infty}^{\infty} \gamma(h)\, e^{-2\pi i \omega h}.$$

This is the **Wiener–Khinchin theorem** and it is the deep reason the time-domain ($\gamma$) and frequency-domain ($f$) descriptions are equivalent: each is the Fourier transform of the other.

### Reading the periodogram

Different stochastic processes leave different signatures in the periodogram:

- **White noise:** flat spectrum, $f(\omega) = \sigma^2$ — equal power at all frequencies (hence "white").
- **AR(1) with $\phi > 0$ ("red noise"):** spectrum decaying with frequency — power concentrated at low frequencies, characteristic of slow, persistent variability. Many astrophysical processes (AGN variability, stellar granulation) look like this.
- **Periodic signal:** a sharp spike at the signal frequency (and at its harmonics if the waveform is non-sinusoidal).
- **Quasi-periodic oscillation (QPO):** a broad bump, narrower than the continuum but wider than a delta function — characteristic of damped oscillators (accreting compact objects show these prominently).

This is why FFT/periodogram analysis is the **first thing you do** in time-domain astronomy: a glance at the spectrum tells you whether you're looking at noise, persistent variability, periodicity, or something more exotic.

### Practical considerations

The naive FFT works beautifully for clean, evenly-spaced, complete data — and quite badly outside that regime. The main practical issues:

1. **Leakage.** A finite-duration observation implicitly multiplies the underlying process by a rectangular window. In the frequency domain this convolves the true spectrum with the window's transform (a sinc function), spreading power from each true frequency into neighboring bins. The fix is **windowing**: multiply the series by a tapering function (Hann, Hamming, Tukey, ...) that goes smoothly to zero at the edges, suppressing leakage at the cost of slightly reduced frequency resolution.

2. **Variance of the periodogram.** Crucially, the periodogram is *not* a consistent estimator of the PSD — its variance does not decrease as $N \to \infty$. Each $I(\omega_k)$ has $\sim 100\%$ relative error regardless of how much data you have. To get a consistent estimate you must **smooth** (average $I$ across neighboring frequencies) or **segment-and-average** (split the series into chunks, periodogram each, average — Welch's method). This is non-negotiable: a single raw periodogram is a noisy snapshot, not a reliable spectrum.

3. **Irregular sampling.** The DFT assumes evenly-spaced data. For unevenly-sampled astronomical light curves the standard substitute is the **Lomb–Scargle periodogram**, which is essentially a least-squares fit of sinusoids at each frequency and reduces to the standard periodogram when the sampling is uniform. We'll meet this later.

4. **Gaps.** Missing data within an otherwise uniform grid creates spectral artifacts (the "window function" of the gaps imprints itself on the spectrum, producing spurious peaks and sidelobes). Mitigations: fill with the mean (crude), interpolate (introduces correlated noise), use Lomb–Scargle on the available points, or fit in the time domain with a GP and Fourier-transform the inferred process.

5. **Mean and trend.** The DFT at $k=0$ is just $N\bar{x}$, dominated by the mean. Always subtract the mean before transforming. If there is a slow trend (linear, polynomial), it leaks power into all low frequencies — detrend first, or accept that the lowest few frequency bins are unreliable.

## Two specialized tools: CARMA and Bayesian Blocks

The classical ARMA framework assumes evenly-spaced, complete data — exactly what astronomical time series typically are *not*. Real light curves arrive irregularly, with gaps, heteroscedastic error bars, and sometimes as event lists (individual photons) rather than fluxes. Two tools widely used in astronomy address these realities head-on: **CARMA** for irregularly-sampled continuous variability, and **Bayesian Blocks** for event data or abrupt level changes.

### CARMA: Continuous-time ARMA

#### Motivation

ARMA models live on a regular integer index $t \in \mathbb{Z}$. If your data are sampled irregularly — as ground-based light curves almost always are — then "lag $h = 1$" has no fixed physical meaning, and the autoregressive recursion $X_t = \phi X_{t-1} + Z_t$ is ill-defined. The natural fix is to write the model in **continuous time** so that any sampling pattern can be accommodated.

#### Definition

A **CARMA($p, q$)** process is the continuous-time analog of ARMA. It is the stationary solution of the stochastic differential equation

$$\frac{d^p X(t)}{dt^p} + \alpha_{p-1}\frac{d^{p-1} X(t)}{dt^{p-1}} + \cdots + \alpha_0 X(t) = \beta_0\, \epsilon(t) + \beta_1 \frac{d\epsilon(t)}{dt} + \cdots + \beta_q\frac{d^q \epsilon(t)}{dt^q},$$

where $\epsilon(t)$ is continuous-time white noise (formally the derivative of a Wiener process) and $\alpha_j, \beta_j$ are the AR and MA coefficients in continuous time. The simplest non-trivial case, **CAR(1)** (also called the **damped random walk**, DRW, or Ornstein–Uhlenbeck process), is

$$dX(t) = -\frac{1}{\tau}\, X(t)\, dt + \sigma\, dW(t),$$

a process that relaxes toward zero on a characteristic timescale $\tau$ while being kicked by Gaussian noise. Higher-order CARMA can produce damped oscillations (CARMA(2,0), CARMA(2,1)) — the natural model for a stochastically-driven, damped harmonic oscillator. CARMA(2,1) is in fact the model behind the **SHO** ("simple harmonic oscillator") kernel that's popular in GP-based light-curve modeling (see `celerite`).

#### Why CARMA fits astronomy

Three big advantages over discrete-time ARMA:

1. **Irregular sampling is natural.** The model is defined for all $t \in \mathbb{R}$; the observation times $t_1, t_2, \ldots$ can be anywhere. The likelihood is computed via the Kalman filter on the irregular grid in $\mathcal{O}(N)$ time.
2. **Heteroscedastic measurement errors are built in.** Each observation can carry its own error bar $\sigma_i$, which enters the Kalman filter cleanly. This matters a lot in astronomy where photometric uncertainties vary from point to point.
3. **Physical interpretation.** The CARMA parameters correspond to relaxation timescales and (possibly) oscillation frequencies of the underlying stochastic process — quantities with direct astrophysical meaning (e.g., the variability timescale $\tau$ of an AGN, related to the size of the emitting region).

#### Connection to the power spectrum

The CARMA PSD has a closed form:

$$f(\omega) = \frac{\sigma^2}{2\pi}\, \frac{\left|\sum_{j=0}^{q} \beta_j (i\omega)^j \right|^2}{\left|\sum_{j=0}^{p} \alpha_j (i\omega)^j\right|^2}, \quad \alpha_p \equiv 1.$$

This is a **rational function of $\omega$** — a ratio of polynomials. For the DRW (CAR(1)), this becomes the classic Lorentzian:

$$f(\omega) \propto \frac{1}{\omega^2 + (1/\tau)^2},$$

flat for $\omega \ll 1/\tau$ and falling as $\omega^{-2}$ for $\omega \gg 1/\tau$ — the canonical "broken power-law" shape seen in AGN light curves. Higher-order CARMA can reproduce more complex broken-power-law and QPO-bearing spectra without needing infinite-parameter models. This is the main reason CARMA has become a standard tool for AGN variability, X-ray binary timing, and stellar microvariability.

#### Practical notes

- Estimation is by maximum likelihood or MCMC using the Kalman-filter likelihood (e.g., `EzTao`, `celerite2`, `Carma_pack`).
- Model order $(p, q)$ selection is non-trivial; standard practice is to fit a range of orders and use AIC/BIC or posterior model probabilities. Over-fitting in $q$ is a known failure mode.
- Stationarity requires the roots of the AR polynomial to lie in the left half-plane (the continuous-time analog of "outside the unit circle").

### Bayesian Blocks

#### Motivation

CARMA assumes the variability is a smooth, stationary stochastic process. The opposite extreme — popular in high-energy astronomy — is when the signal consists of distinct **piecewise-constant epochs** separated by abrupt changes: a gamma-ray burst light curve with distinct emission episodes, an X-ray source flipping between high and low states, a flare on top of quiescence. For these, the natural question is not "what's the spectrum?" but "where are the change points, and what's the rate in each segment?".

**Bayesian Blocks** (Scargle 1998, 2013) is a nonparametric method to find the optimal segmentation of a time series into piecewise-constant blocks. It is fundamentally a model-selection algorithm: every possible partition is a model, and we pick the partition with the best penalized likelihood.

#### What it does

Given a sequence of events (photon arrival times) or measurements, Bayesian Blocks finds the partition of the observation interval into $K$ contiguous segments, with $K$ itself determined by the data, that maximizes

$$\log L_{\rm total} = \sum_{k=1}^{K} \log L_k - K \cdot \text{ncp\_prior},$$

where $L_k$ is the likelihood of segment $k$ under a constant-rate (or constant-mean) model, and `ncp_prior` is a penalty per change point that controls the false-alarm rate. The "block representation" of the data is a step function: constant within each block, jumping at the change points.

Three data modes are supported, each with its own per-block likelihood:

1. **Event data** (photon arrival times, points on the time axis): Poisson likelihood, block rate = (events in block) / (block duration).
2. **Binned counts** (counts in pre-defined bins): Poisson likelihood for binned data.
3. **Measurements with errors** (point-measurement mode): Gaussian likelihood, block mean = inverse-variance-weighted mean of points in block.

#### Why it works well

- **Optimal in a precise sense.** Despite the combinatorial explosion of possible partitions ($2^{N-1}$ for $N$ data points), there is an $\mathcal{O}(N^2)$ dynamic programming algorithm that finds the *globally* optimal partition. You don't get stuck in local optima.
- **Adaptive resolution.** Block widths are determined by the data: short blocks where variability is rapid, long blocks where the source is quiet. This is exactly what you want for a flaring source observed at fixed cadence.
- **Handles irregular sampling and exposure variations naturally** through the per-block likelihood.
- **No bin-width parameter to tune.** Unlike a histogram, you don't have to choose $\Delta t$; the algorithm chooses the segmentation for you.
- **`ncp_prior` calibration is principled.** The prior penalty can be tuned to give a desired false-positive rate for change-point detection (e.g., one false change point per realization), and standard formulas in Scargle+ 2013 give appropriate values as a function of $N$.

#### Typical use cases

- **GRB light curves:** identifying emission episodes, separating prompt phases from afterglow rises.
- **AGN flares:** finding the onset and end of a flare against a noisy baseline.
- **State transitions in X-ray binaries:** automatically segmenting hard-state and soft-state intervals.
- **Constructing adaptively-binned light curves** for visualization and follow-up analysis, instead of the fixed-bin histograms that hide narrow features.

Implementation is in `astropy.stats.bayesian_blocks` and is essentially a one-liner once you have the data.

In practice you often want both: use Bayesian Blocks to detect and isolate distinct epochs of activity, then fit a CARMA (or GP) model to the variability *within* the quiescent or flaring blocks. The two are complementary, not competing. -->

$$\begin{aligned}
X_t &= \phi X_{t-1} + Z_t \
&= \phi(\phi X_{t-2} + Z_{t-1}) + Z_t = \phi^2 X_{t-2} + \phi Z_{t-1} + Z_t \
&= \phi^2(\phi X_{t-3} + Z_{t-2}) + \phi Z_{t-1} + Z_t = \phi^3 X_{t-3} + \phi^2 Z_{t-2} + \phi Z_{t-1} + Z_t \
&;;\vdots \
&= \phi^k X_{t-k} + \sum_{j=0}^{k-1} \phi^j Z_{t-j}.
\end{aligned}$$
